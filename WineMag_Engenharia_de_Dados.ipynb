**Confiabilidade das avaliações** Assume-se que as pontuações atribuídas aos vinhos são consistentes e confiáveis, uma vez que foram produzidas por especialistas da revista Wine Enthusiast.

**Preço como indicador de valor percebido** Considera-se que o preço informado representa o valor médio de mercado no momento da avaliação e pode ser utilizado como proxy de valor percebido pelo consumidor.

**Qualidade representada pela pontuação** A pontuação (points) é tratada como indicador direto de qualidade do vinho, permitindo comparações entre países, uvas e regiões.

**Integridade das informações categóricas** As categorias de país, variedade de uva, região e vinícola são assumidas como corretas conforme disponibilizadas na fonte original.

**Tratamento de dados ausentes** Registros com ausência de preço ou outras variáveis relevantes poderão ser tratados, imputados ou excluídos conforme necessidade analítica, sem comprometer o objetivo do estudo.

**Escopo limitado à análise estrutural** A análise concentra-se nos atributos estruturados do dataset, não abrangendo processamento avançado das descrições textuais (NLP), exceto quando necessário para contextualização.

**Representatividade do dataset** Reconhece-se que o conjunto de dados não representa todo o mercado global de vinhos, mas apenas os rótulos avaliados pela revista, podendo haver vieses geográficos e de seleção.

# 5.2 Extração dos Dados (E)

A extração consiste na leitura do arquivo armazenado na camada Bronze. Nesta etapa:

O arquivo CSV é carregado para um DataFrame no Databricks.

O schema original é inspecionado.

São verificadas colunas, tipos e volume de registros.

Nenhuma modificação é aplicada nesta fase.

Essa leitura inicial prepara o dataset para as transformações subsequentes.

```python
# 1. Garantir catálogo e schema corretos
spark.sql("USE CATALOG winemag_catalog")
spark.sql("USE SCHEMA bronze")

# 2. Ler o CSV corretamente (separador ;)
df = (
    spark.read.format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .option("delimiter", ";")
        .load("/Volumes/winemag_catalog/bronze/volume/winemag-data-130k-v2.csv")
)

display(df)
df.printSchema()

# 3. Normalizar nomes das colunas
import unicodedata
import re

def normalize(col):
    col = col.strip()
    col = unicodedata.normalize('NFKD', col).encode('ascii', 'ignore').decode('ascii')
    col = col.lower()
    col = re.sub(r'[^a-zA-Z0-9_]', '_', col)
    col = re.sub(r'_+', '_', col)
    return col

df = df.toDF(*[normalize(c) for c in df.columns])

# 4. Criar a tabela Bronze winemag_raw
df.write.format("delta").mode("overwrite").saveAsTable(
    "winemag_catalog.bronze.winemag_raw"
)

# 5. Validar
display(spark.table("winemag_catalog.bronze.winemag_raw"))
```


Step 05: Análise inicial de outliers

```python
#Estatísticas descritivas
df.describe(["price", "points"]).show()
```

```python
#Boxplot (Databricks display)
display(df.select("preco"))
```

```python
#Outliers de preço (IQR)

from pyspark.sql.functions import col, expr

# 1. Criar coluna numérica segura
df2 = df.withColumn("preco_float", expr("try_cast(preco as float)"))

# 2. Filtrar apenas linhas onde preco é realmente numérico
df_num = df2.filter(col("preco_float").isNotNull())

# 3. Calcular Q1 e Q3 usando apenas valores numéricos
q1 = df_num.approxQuantile("preco_float", [0.25], 0.05)[0]
q3 = df_num.approxQuantile("preco_float", [0.75], 0.05)[0]

# 4. Calcular IQR
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# 5. Filtrar outliers apenas entre valores numéricos
outliers_price = df_num.filter(
    (col("preco_float") < limite_inferior) |
    (col("preco_float") > limite_superior)
)

display(outliers_price)
```

```python
#Outliers de pontuação
q1 = df.approxQuantile("pontos", [0.25], 0.05)[0]
q3 = df.approxQuantile("pontos", [0.75], 0.05)[0]
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers_points = df.filter(
    (col("pontos") < limite_inferior) | (col("pontos") > limite_superior)
)

display(outliers_points)
```

Preencher nulos em colunas numéricas com média


```python
from pyspark.sql.functions import avg

#Preencher nulos em colunas numéricas com média
media_price = df.select(avg("preco")).first()[0]
df = df.fillna({"preco": media_price})

#Preencher nulos em colunas categóricas com “Unknown”
categoricas = [
    "designacao", "estado", "regiao", "regiao2",   "nome_avaliador", "perfil",  "nome_do_vinho","variedade","vinicula"]
df = df.fillna({c: "Unknown" for c in categoricas})
```

```python
#Converter colunas numéricas

from pyspark.sql.functions import col
df = df.withColumn("pontos", col("pontos").cast("int"))
df = df.withColumn("pontos", col("pontos").cast("double"))

#Garantir que colunas de texto são string
texto_cols = ["descricao", "pais", "variedade", "vinicula", "nome_do_vinho"]
for c in texto_cols:
    df = df.withColumn(c, col(c).cast("string"))
```


Step 04: Normalização de textos



```python
#Remover espaços extras e padronizar para minúsculas

from pyspark.sql.functions import trim, lower
for c in texto_cols:
    df = df.withColumn(c, lower(trim(col(c))))
```

```python
#Remover caracteres especiais de colunas categóricas

from pyspark.sql.functions import regexp_replace
for c in texto_cols:
    df = df.withColumn(c, regexp_replace(col(c), r"[^a-zA-Z0-9\s]", ""))
```

```python
#Estatísticas descritivas (base para detectar outliers)
df.describe(["price", "points"]).show()

```

```python
#Cálculo de limites usando IQR (Método mais usado)

from pyspark.sql.functions import col

# Preço
q1_price = df.approxQuantile("price", [0.25], 0.05)[0]
q3_price = df.approxQuantile("price", [0.75], 0.05)[0]
iqr_price = q3_price - q1_price

lim_inf_price = q1_price - 1.5 * iqr_price
lim_sup_price = q3_price + 1.5 * iqr_price

# Pontuação
q1_points = df.approxQuantile("points", [0.25], 0.05)[0]
q3_points = df.approxQuantile("points", [0.75], 0.05)[0]
iqr_points = q3_points - q1_points

lim_inf_points = q1_points - 1.5 * iqr_points
lim_sup_points = q3_points + 1.5 * iqr_points

```

```python
#Identificar outliers
outliers_price = df.filter((col("price") < lim_inf_price) | (col("price") > lim_sup_price))
outliers_points = df.filter((col("points") < lim_inf_points) | (col("points") > lim_sup_points))

display(outliers_price)
display(outliers_points)
```

```python
#Remover outliers (opcional)

df = df.filter(
    (col("price") >= lim_inf_price) & (col("price") <= lim_sup_price) &
    (col("points") >= lim_inf_points) & (col("points") <= lim_sup_points)
)
```

#5.3.4 Preparação para Modelagem Dimensional

A preparação para modelagem dimensional consiste em transformar os dados tratados (Silver) em estruturas analíticas otimizadas (Gold), seguindo o padrão Kimball de Data Warehousing.

O objetivo é:

- separar entidades
- criar dimensões
- criar chaves substitutas
- garantir integridade referencial
- montar a tabela fato
- gerar o modelo estrela final


#Step 01: Separação das entidades

Nesta etapa identificamos quais atributos representam dimensões e quais representam fatos.

Dimensões identificadas:
País / Região

Vinícola (Winery)

Variedade (Grape Variety)

Avalista (Taster)

Vinho (Wine)

Fato identificada:
Avaliação do Vinho (nota, preço, descrição, vínculos dimensionais)

```python
df = spark.table("winemag_catalog.silver.winemag_semantic")

dim_pais_regiao = df.select("country", "province", "region_1", "region_2").dropDuplicates()
dim_vinicula    = df.select("winery").dropDuplicates()
dim_variedade   = df.select("variety").dropDuplicates()
dim_avalista    = df.select("taster_name").dropDuplicates()
dim_vinho       = df.select("title", "variety", "winery").dropDuplicates()
```

#Step 02: Criação de chaves substitutas

Chaves substitutas são essenciais para:

garantir integridade

evitar dependência de chaves naturais

permitir histórico (SCD) no futuro

```python
from pyspark.sql.functions import monotonically_increasing_id

dim_pais_regiao = dim_pais_regiao.withColumn("id_pais_regiao", monotonically_increasing_id())
dim_vinicula    = dim_vinicula.withColumn("id_vinicula", monotonically_increasing_id())
dim_variedade   = dim_variedade.withColumn("id_variedade", monotonically_increasing_id())
dim_avalista    = dim_avalista.withColumn("id_avalista", monotonically_increasing_id())
dim_vinho       = dim_vinho.withColumn("id_vinho", monotonically_increasing_id())
```

#Step 03: Normalização das tabelas

Cada dimensão é salva no schema gold.

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS winemag_catalog.gold")

dim_pais_regiao.write.format("delta").mode("overwrite").saveAsTable("winemag_catalog.gold.dim_pais_regiao")
dim_vinicula.write.format("delta").mode("overwrite").saveAsTable("winemag_catalog.gold.dim_vinicula")
dim_variedade.write.format("delta").mode("overwrite").saveAsTable("winemag_catalog.gold.dim_variedade")
dim_avalista.write.format("delta").mode("overwrite").saveAsTable("winemag_catalog.gold.dim_avalista")
dim_vinho.write.format("delta").mode("overwrite").saveAsTable("winemag_catalog.gold.dim_vinho")
```

Step 04: Garantia de integridade referencial

Agora fazemos os joins entre a Silver e as dimensões para montar a fato.

```python
df = spark.table("winemag_catalog.silver.winemag_semantic")

dim_pais_regiao = spark.table("winemag_catalog.gold.dim_pais_regiao")
dim_vinicula    = spark.table("winemag_catalog.gold.dim_vinicula")
dim_variedade   = spark.table("winemag_catalog.gold.dim_variedade")
dim_avalista    = spark.table("winemag_catalog.gold.dim_avalista")
dim_vinho       = spark.table("winemag_catalog.gold.dim_vinho")

fact = df \
    .join(dim_pais_regiao, ["country", "province", "region_1", "region_2"], "left") \
    .join(dim_vinicula, ["winery"], "left") \
    .join(dim_variedade, ["variety"], "left") \
    .join(dim_avalista, ["taster_name"], "left") \
    .join(dim_vinho, ["title", "variety", "winery"], "left")
```


Step 05: Geração da Tabela Fato

Selecionamos:

chaves substitutas
métricas
atributos descritivos úteis

```python
from pyspark.sql.functions import col

fact_avaliacao_vinho = fact.select(
    col("id_vinho"),
    col("id_vinicula"),
    col("id_variedade"),
    col("id_pais_regiao"),
    col("id_avalista"),
    col("points").alias("nota"),
    col("price").alias("preco"),
    col("description").alias("descricao"),
    col("designation").alias("designacao"),
    col("taster_name"),
    col("taster_twitter_handle"),
    col("country"),
    col("province"),
    col("region_1"),
    col("region_2"),
    col("winery"),
    col("title"),
    col("variety")
)

fact_avaliacao_vinho.write.format("delta").mode("overwrite").saveAsTable(
    "winemag_catalog.gold.fact_avaliacao_vinho"
)
```

#Step 04: Garantia de integridade referencial (montar a fato)

Agora vamos “voltar” a chave substituta para o fato, fazendo joins da Silver com as dimensões.



```python
python
from pyspark.sql.functions import col

# recarregar dimensões (garante uso da versão salva)
dim_pais_regiao = spark.table("winemag_catalog.gold.dim_pais_regiao")
dim_vinicula    = spark.table("winemag_catalog.gold.dim_vinicula")
dim_variedade   = spark.table("winemag_catalog.gold.dim_variedade")
dim_avalista    = spark.table("winemag_catalog.gold.dim_avalista")
dim_vinho       = spark.table("winemag_catalog.gold.dim_vinho")

# começa da silver
fact = df

# join com dim_pais_regiao
fact = (
    fact.join(
        dim_pais_regiao,
        on=["country", "province", "region_1", "region_2"],
        how="left"
    )
)

# join com dim_vinicula
fact = (
    fact.join(
        dim_vinicula,
        on=["winery"],
        how="left"
    )
)

# join com dim_variedade
fact = (
    fact.join(
        dim_variedade,
        on=["variety"],
        how="left"
    )
)

# join com dim_avalista
fact = (
    fact.join(
        dim_avalista,
        on=["taster_name"],
        how="left"
    )
)

# join com dim_vinho
fact = (
    fact.join(
        dim_vinho,
        on=["title", "variety", "winery"],
        how="left"
    )
)
```

#Step 05: Fato final e gravação em Gold
Vamos escolher as colunas da fato: chaves + métricas + alguns atributos descritivos.



```python
from pyspark.sql.functions import col

fact_avaliacao_vinho = fact.select(
    # chaves substitutas
    col("id_vinho"),
    col("id_vinicula"),
    col("id_variedade"),
    col("id_pais_regiao"),
    col("id_avalista"),

    # métricas
    col("points").alias("nota"),
    col("price").alias("preco"),

    # atributos descritivos úteis
    col("description").alias("descricao"),
    col("designation").alias("designacao"),
    col("taster_name"),
    col("taster_twitter_handle"),
    col("country"),
    col("province"),
    col("region_1"),
    col("region_2"),
    col("winery"),
    col("title"),
    col("variety")
)

# grava tabela fato em Gold
fact_avaliacao_vinho.write.format("delta").mode("overwrite").saveAsTable(
    "winemag_catalog.gold.fact_avaliacao_vinho"
)

display(fact_avaliacao_vinho.limit(20))
```
