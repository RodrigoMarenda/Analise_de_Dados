# 5.2 Extração dos Dados (E)

A extração consiste na leitura do arquivo armazenado na camada Bronze.

Nessa etapa, o arquivo CSV é carregado para um DataFrame no Databricks. Nenhuma modificação é aplicada nesta fase.

Essa leitura inicial prepara o dataset para as modificações na etapa 5.3 Transformação dos Dados (T).

```python
# 1. Garantir catálogo e schema corretos
spark.sql("USE CATALOG winemag_catalog")
spark.sql("USE SCHEMA bronze")

# 2. Ler o CSV corretamente (separador ;)
df = (
    spark.read.format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .option("delimiter", ";")
        .load("/Volumes/winemag_catalog/bronze/volume/winemag-data-130k-v2.csv")
)

display(df)
df.printSchema()

# 3. Normalizar nomes das colunas
import unicodedata
import re

def normalize(col):
    col = col.strip()
    col = unicodedata.normalize('NFKD', col).encode('ascii', 'ignore').decode('ascii')
    col = col.lower()
    col = re.sub(r'[^a-zA-Z0-9_]', '_', col)
    col = re.sub(r'_+', '_', col)
    return col

df = df.toDF(*[normalize(c) for c in df.columns])

# 4. Criar a tabela Bronze winemag_raw
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("winemag_catalog.bronze.winemag_raw")

# 5. Validar
display(spark.table("winemag_catalog.bronze.winemag_raw"))
```

```python
#Outliers de preço (IQR)

from pyspark.sql.functions import col, expr

# 1. Criar coluna numérica segura
df2 = df.withColumn("preco_float", expr("try_cast(preco as float)"))

# 2. Filtrar apenas linhas onde preco é realmente numérico
df_num = df2.filter(col("preco_float").isNotNull())

# 3. Calcular Q1 e Q3 usando apenas valores numéricos
q1 = df_num.approxQuantile("preco_float", [0.25], 0.05)[0]
q3 = df_num.approxQuantile("preco_float", [0.75], 0.05)[0]

# 4. Calcular IQR
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# 5. Filtrar outliers apenas entre valores numéricos
outliers_price = df_num.filter(
    (col("preco_float") < limite_inferior) |
    (col("preco_float") > limite_superior)
)

display(outliers_price)
```

![Evid_5.3.1_Step05.3.png]

```python
#Outliers de pontuação
from pyspark.sql.functions import col, expr

# 1. Criar coluna numérica segura para pontos
df2 = df.withColumn("pontos_float", expr("try_cast(pontos as double)"))
# 2. Filtrar apenas linhas onde pontos são realmente numéricos
df_num = df2.filter(col("pontos_float").isNotNull())
# 3. Calcular Q1 e Q3 usando apenas valores válidos
q1 = df_num.approxQuantile("pontos_float", [0.25], 0.05)[0]
q3 = df_num.approxQuantile("pontos_float", [0.75], 0.05)[0]
# 4. Calcular IQR
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr
# 5. Filtrar outliers apenas entre valores numéricos
outliers_points = df_num.filter(
    (col("pontos_float") < limite_inferior) |
    (col("pontos_float") > limite_superior)
)
display(outliers_points)
```

```python
from pyspark.sql.functions import avg

#Preencher nulos em colunas numéricas com média
media_price = df.select(avg("preco")).first()[0]
df = df.fillna({"preco": media_price})

#Preencher nulos em colunas categóricas com “Unknown”
categoricas = [
    "designacao", "estado", "regiao", "regiao2",   "nome_avaliador", "perfil",  "nome_do_vinho","variedade","vinicula"]
df = df.fillna({c: "Unknown" for c in categoricas})
```

Step 03: Padronização de tipos


```python
# Imports necessários
from pyspark.sql.functions import expr, col

# Converter colunas numéricas de forma segura
for c in ["preco", "pontos"]:
    df = df.withColumn(c, expr(f"try_cast({c} as double)"))

# Garantir que colunas de texto são string
texto_cols = ["descricao", "pais", "variedade", "vinicula", "nome_do_vinho"]
for c in texto_cols:
    df = df.withColumn(c, col(c).cast("string"))

# Conferir o schema atualizado
df.printSchema()
```

```python
#Remover espaços extras e padronizar para minúsculas

from pyspark.sql.functions import trim, lower
for c in texto_cols:
    df = df.withColumn(c, lower(trim(col(c))))
```

```python
#Remover caracteres especiais de colunas categóricas

from pyspark.sql.functions import regexp_replace
for c in texto_cols:
    df = df.withColumn(c, regexp_replace(col(c), r"[^a-zA-Z0-9\s]", ""))
```

calcula os limites de valores aceitáveis para preco e pontos usando estatística (IQR), para depois filtrar ou identificar outliers

```python
#calcula os limites de valores aceitáveis para preco e pontos usando estatística (IQR),
# para depois filtrar ou identificar outliers

from pyspark.sql.functions import col, expr

# 1. Converter preco para float de forma segura
df2 = df.withColumn("preco_float", expr("try_cast(preco as double)"))

# 2. Filtrar apenas linhas onde preco é numérico
df_num_price = df2.filter(col("preco_float").isNotNull())

# --- Preço ---
q1_price = df_num_price.approxQuantile("preco_float", [0.25], 0.05)[0]
q3_price = df_num_price.approxQuantile("preco_float", [0.75], 0.05)[0]
iqr_price = q3_price - q1_price

lim_inf_price = q1_price - 1.5 * iqr_price
lim_sup_price = q3_price + 1.5 * iqr_price

# --- Pontuação ---
# pontos já está como double no seu schema, então não precisa converter
df_num_points = df.filter(col("pontos").isNotNull())

q1_points = df_num_points.approxQuantile("pontos", [0.25], 0.05)[0]
q3_points = df_num_points.approxQuantile("pontos", [0.75], 0.05)[0]
iqr_points = q3_points - q1_points

lim_inf_points = q1_points - 1.5 * iqr_points
lim_sup_points = q3_points + 1.5 * iqr_points


```

![Evid_5.3.3_Step01.2.png]

```python
#Identificar outliers
outliers_price = df.filter((col("preco") < lim_inf_price) | (col("preco") > lim_sup_price))
outliers_points = df.filter((col("pontos") < lim_inf_points) | (col("pontos") > lim_sup_points))

display(outliers_price)
display(outliers_points)
```

```python
#Remover outliers

# Definir limites manualmente
lim_inf_price = 0.0
lim_sup_price = 500.0
lim_inf_points = 80
lim_sup_points = 100

# Aplicar filtro
df = df.filter(
    (col("preco") >= lim_inf_price) & (col("preco") <= lim_sup_price) &
    (col("pontos") >= lim_inf_points) & (col("pontos") <= lim_sup_points)
)
```

#5.3.4 Preparação para Modelagem Dimensional

A preparação para modelagem dimensional consiste em transformar os dados tratados (Silver) em estruturas analíticas otimizadas (Gold).

O objetivo é:

- separar entidades
- criar dimensões
- criar chaves substitutas
- garantir integridade referencial

#Step 01: Separação das entidades

Nesta etapa vamos identificar quais atributos representam dimensões e quais representam fatos.

Dimensões identificadas:

- País / Região
- Vinícola
- Variedade
- Avalista
- Vinho

Fato identificada:

Chaves substitutas para garantir integridade, evitar dependência de chaves naturais e permitir histórico no futuro.

# Step 04: Garantia de integridade referencial

Agora fazemos os joins entre a Silver e as dimensões para montar a fato.

```python
df = spark.table("winemag_catalog.silver.winemag_semantic")

dim_pais_regiao = spark.table("winemag_catalog.gold.dim_pais_regiao")
dim_vinicula    = spark.table("winemag_catalog.gold.dim_vinicula")
dim_variedade   = spark.table("winemag_catalog.gold.dim_variedade")
dim_avalista    = spark.table("winemag_catalog.gold.dim_avalista")
dim_vinho       = spark.table("winemag_catalog.gold.dim_vinho")

fato = df \
    .join(dim_pais_regiao, ["pais", "estado", "regiao", "regiao2"], "left") \
    .join(dim_vinicula, ["vinicula"], "left") \
    .join(dim_variedade, ["variedade"], "left") \
    .join(dim_avalista, ["nome_avaliador"], "left") \
    .join(dim_vinho, ["nome_do_vinho", "variedade", "vinicula"], "left")

```

Agora vamos “voltar” a chave substituta para o fato, fazendo joins da Silver com as dimensões.



```python
from pyspark.sql.functions import col

# recarregar dimensões (garante uso da versão salva)
dim_pais_regiao = spark.table("winemag_catalog.gold.dim_pais_regiao")
dim_vinicula    = spark.table("winemag_catalog.gold.dim_vinicula")
dim_variedade   = spark.table("winemag_catalog.gold.dim_variedade")
dim_avalista    = spark.table("winemag_catalog.gold.dim_avalista")
dim_vinho       = spark.table("winemag_catalog.gold.dim_vinho")

# começa da silver
fato = df

# join com dim_pais_regiao
fato = (
    fato.join(
        dim_pais_regiao,
        on=["pais", "estado", "regiao", "regiao2"],
        how="left"))

# join com dim_vinicula
fato = (
    fato.join(
        dim_vinicula,
        on=["vinicula"],
        how="left"))

# join com dim_variedade
fato = (
    fato.join(
        dim_variedade,
        on=["variedade"],
        how="left"))

# join com dim_avalista
fato = (
    fato.join(
        dim_avalista,
        on=["nome_avaliador"],
        how="left"))

# join com dim_vinho
fato = (
    fato.join(
        dim_vinho,
        on=["nome_do_vinho", "variedade", "vinicula"],
        how="left"))
```

#STEP 01: Qualidade dos Dados

1. COMPLETUDE (valores nulos)

```python
%sql
#============================
# 1. COMPLETUDE (valores nulos)
# ============================
SELECT 'nulos_nota' TITULO, COUNT(*) AS nulos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE nota IS NULL
UNION
SELECT 'nulos_preco', COUNT(*) AS nulos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE preco IS NULL
UNION
SELECT 'nulos_descricao' , COUNT(*) AS nulos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE descricao IS NULL
UNION
SELECT 'nulos_designacao' , COUNT(*) AS nulos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE designacao IS NULL;
```

![Evid_6_Step01.1.png]

2. CONSISTÊNCIA (FKs válidos)


```python
%sql
# ============================
# 2. CONSISTÊNCIA (FKs válidos)
# ============================
SELECT 'fk_sem_vinho' TITULO, COUNT(*) AS inconsistencias
FROM winemag_catalog.gold.fato_avaliacao_vinho f
LEFT JOIN winemag_catalog.gold.dim_vinho v
  ON f.id_vinho = v.id_vinho
WHERE v.id_vinho IS NULL
UNION
SELECT 'fk_sem_variedade', COUNT(*) AS inconsistencias
FROM winemag_catalog.gold.fato_avaliacao_vinho f
LEFT JOIN winemag_catalog.gold.dim_variedade d
  ON f.id_variedade = d.id_variedade
WHERE d.id_variedade IS NULL
UNION
SELECT 'fk_sem_vinicula', COUNT(*) AS inconsistencias
FROM winemag_catalog.gold.fato_avaliacao_vinho f
LEFT JOIN winemag_catalog.gold.dim_vinicula d
  ON f.id_vinicula = d.id_vinicula
WHERE d.id_vinicula IS NULL
UNION
SELECT 'fk_sem_pais_regiao', COUNT(*) AS inconsistencias
FROM winemag_catalog.gold.fato_avaliacao_vinho f
LEFT JOIN winemag_catalog.gold.dim_pais_regiao d
  ON f.id_pais_regiao = d.id_pais_regiao
WHERE d.id_pais_regiao IS NULL
UNION
SELECT 'fk_sem_avalista', COUNT(*) AS inconsistencias
FROM winemag_catalog.gold.fato_avaliacao_vinho f
LEFT JOIN winemag_catalog.gold.dim_avalista d
  ON f.id_avalista = d.id_avalista
WHERE d.id_avalista IS NULL;
```

![Evid_6_Step01.2.png]

3. UNICIDADE (PKs sem repetição)


```python
%sql
# ============================
# 3. UNICIDADE (PKs sem repetição)
# ============================
SELECT 'duplicados_dim_vinho' TITULO, COUNT(*) AS repeticoes
FROM (
    SELECT id_vinho
    FROM winemag_catalog.gold.dim_vinho
    GROUP BY id_vinho
    HAVING COUNT(*) > 1
) t
UNION
SELECT 'duplicados_dim_variedade', COUNT(*) AS repeticoes
FROM (
    SELECT id_variedade
    FROM winemag_catalog.gold.dim_variedade
    GROUP BY id_variedade
    HAVING COUNT(*) > 1
) t;
```

![Evid_6_Step01.3.png]

4. VALIDADE (intervalos e formatos)

```python
%sql
# ============================
# 4. VALIDADE (intervalos e formatos)
# ============================
SELECT 'notas_invalidas' TITULO, COUNT(*) AS invalidos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE nota < 0 OR nota > 100
UNION
SELECT 'precos_invalidos', COUNT(*) AS invalidos
FROM winemag_catalog.gold.fato_avaliacao_vinho
WHERE preco < 0;
```

![Evid_6_Step01.4.png]

5. INTEGRIDADE REFERENCIAL

```python
%sql
# ============================
# 5. INTEGRIDADE REFERENCIAL
# ============================
SELECT f.nota, f.preco, v.nome_do_vinho, a.nome_avaliador, pr.pais
FROM winemag_catalog.gold.fato_avaliacao_vinho f
JOIN winemag_catalog.gold.dim_vinho v ON f.id_vinho = v.id_vinho
JOIN winemag_catalog.gold.dim_avalista a ON f.id_avalista = a.id_avalista
JOIN winemag_catalog.gold.dim_pais_regiao pr ON f.id_pais_regiao = pr.id_pais_regiao
LIMIT 20;
```

![Evid_6_Step01.5.png]

#Step 02: Solução do Problema

Pergunta 1: Quais países têm as maiores notas médias de vinho?



```python
%sql SELECT pais, AVG(nota) AS media_nota
FROM winemag_catalog.gold.fato_avaliacao_vinho
GROUP BY pais
ORDER BY media_nota DESC;
```

![Evid_6_Step02.1.png]

Os países com maiores notas médias foram Reino Unido (91,6), Índia (90,2) e Áustria (90,0). Já produtores tradicionais como França, Itália e Portugal ficaram em torno de 88 pontos, enquanto o Brasil aparece com média de 84,7, mais abaixo no ranking.

Isso indica que países com poucas amostras avaliadas (como Reino Unido e Índia) podem ter médias mais altas por concentrarem vinhos de qualidade superior, enquanto grandes produtores têm notas médias mais equilibradas devido à diversidade de rótulos analisados.

Pergunta 2: Existe relação entre preço e nota?


```python
%sql SELECT nota, AVG(preco) AS preco_medio
FROM winemag_catalog.gold.fato_avaliacao_vinho
GROUP BY nota
ORDER BY nota DESC;
```

![Evid_6_Step02.2.png]

Os resultados mostram uma correlação positiva clara: quanto maior a nota média do vinho, maior também o preço médio. Vinhos avaliados com 96 pontos custam em média 54,1, enquanto os de 80 pontos ficam em torno de 16,3.

Isso indica que o mercado valoriza diretamente a qualidade percebida: vinhos com notas mais altas tendem a ser vendidos por preços mais elevados. Para o consumidor, há uma relação custo‑benefício a considerar; para produtores e comerciantes, confirma que investir em qualidade impacta no valor de mercado.

Pergunta 3: Quais vinícolas se destacam em cada região?


```python
%sql SELECT vinicula, regiao, AVG(nota) AS media_nota
FROM winemag_catalog.gold.fato_avaliacao_vinho
GROUP BY vinicula, regiao
ORDER BY media_nota DESC;
```

![Evid_6_Step02.3.png]

Os resultados mostram que várias vinícolas atingiram notas médias muito altas (95–96 pontos) em diferentes regiões. Exemplos incluem dalfonsocurran em Santa Barbara County, conn creek em Atlas Peak, propriet sperino em Lessona, macauley em Alexander Valley e ken wright em Ribbon Ridge. Também se destacam regiões tradicionais como Brunello di Montalcino na Itália e Sauternes na França, com vinícolas de excelência.

Isso evidencia que tanto regiões clássicas quanto terroirs emergentes apresentam vinícolas de altíssima qualidade, reforçando a diversidade e competitividade global na produção de vinhos.
